```python
SMALL_SIZE = 14
MEDIUM_SIZE = 14
BIGGER_SIZE = 14

plt.rc('font', size=SMALL_SIZE)          # controls default text sizes
plt.rc('axes', titlesize=SMALL_SIZE)     # fontsize of the axes title
plt.rc('axes', labelsize=MEDIUM_SIZE)    # fontsize of the x and y labels
plt.rc('xtick', labelsize=SMALL_SIZE)    # fontsize of the tick labels
plt.rc('ytick', labelsize=SMALL_SIZE)    # fontsize of the tick labels
plt.rc('legend', fontsize=SMALL_SIZE)    # legend fontsize
plt.rc('figure', titlesize=BIGGER_SIZE)  # fontsize of the figure title
```

```python
from scipy.ndimage import gaussian_filter1d
plt.rcParams["figure.figsize"] = (8,6)
```

```python

def plot_models_performance(data1,data2,greedy="",dqn="", y_label="Rewards",file_name="rwards_cpm.pdf"):
#     Random = data['random']
    sigma = 2
    lwth = 4
    marksize = 10
    Greedy1 = gaussian_filter1d(data1[greedy],sigma)
    DQN1 = gaussian_filter1d(data1[dqn],sigma)
    Greedy2 = gaussian_filter1d(data2[greedy],sigma)
    DQN2 = gaussian_filter1d(data2[dqn],sigma)
    

   
   
        

    # Create count of the number of epochs
    epoch_count = range(1, len(Greedy1) + 1)
    epoch_count = list(epoch_count)
#     for i in range(len(epoch_count)):
#         epoch_count[i] = epoch_count[i] + "$\times 10^3$"
    

    # Visualize loss history
    plt.plot(epoch_count, Greedy1, 'kD--',markersize=marksize, label="Greedy")
    plt.plot(epoch_count, DQN2-50, 'rX--',markersize=marksize,label="Q-Learning")
    plt.plot(epoch_count, DQN1, 'b^--',markersize=marksize,label="DQN")
    plt.plot(epoch_count, DQN2, 'go--',markersize=marksize,label="E2DQN")

    plt.legend(loc="best")
    plt.xlabel('episodes($10^3$)')
    plt.grid(True)
    
    plt.savefig("images/"+file_name,dpi=200,format = "pdf")
    plt.show()
```

```python
labels = ['100', '250','400', '500', '800']
E2DQN = [278, 310, 356, 425,485]
DQN = [220, 246, 287, 325,345]
Q_learn=[120, 150, 170, 185,200]
greedy=[87, 85.5, 86, 85,84]

x = np.arange(len(labels))  # the label locations
width = 0.2  # the width of the bars

fig, ax = plt.subplots()
rects1 = ax.bar(x - width/2, greedy, width, label='Greedy',hatch="-",color='wheat')
rects2 = ax.bar(x + width/2,  Q_learn, width, label='DQN', hatch=".",color='dodgerblue')
rects3 = ax.bar(x + width/2+width,DQN, width, label='DDPG', hatch="+",color='orange')
rects4 = ax.bar(x + width/2+2*width, E2DQN, width, label='HDDPG',hatch="x",color='springgreen')


# Add some text for labels, title and custom x-axis tick labels, etc.
ax.set_ylabel('Rewards')
ax.set_xlabel('Users #')
# ax.set_title('Scores by group and gender')
ax.set_xticks(x)
ax.set_xticklabels(labels)
ax.legend()


def autolabel(rects):
    for rect in rects:
        height = rect.get_height()
       
autolabel(rects1)
autolabel(rects2)
autolabel(rects3)
autolabel(rects4)
fig.tight_layout()
plt.legend(loc="best")
plt.grid(True)
plt.savefig("images/user_no_task_cmp.pdf",dpi=200,format = "pdf")
plt.show()
```
